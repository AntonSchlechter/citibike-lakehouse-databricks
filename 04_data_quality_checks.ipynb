```python
silver_table = "citibike_lakehouse.silver.trips"

df_silver = spark.table(silver_table)

checks = {
    "row_count": df_silver.count(),
    "null_ride_id": df_silver.filter("ride_id IS NULL").count(),
    "null_started_at": df_silver.filter("started_at IS NULL").count(),
    "invalid_duration": df_silver.filter("ride_duration_minutes <= 0").count(),
    "duplicate_ride_id": df_silver.groupBy("ride_id").count().filter("count > 1").count()
}

if checks["row_count"] == 0:
    raise Exception("Silver Table is empty")

if checks["null_ride_id"] > 0:
    raise Exception("ride_id contains null values")

if checks["null_started_at"] > 0:
    raise Exception("null_started_at contains null values")

if checks["invalid_duration"] > 0
raise Exception("Invalid ride durations found")

if checks["duplicate_ride_id"] > 0:
    raise Exception("Duplicate ride_id values found")
```
